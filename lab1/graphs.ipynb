```python
import sys
import matplotlib.pyplot as plt

def load(filename, start=1):
    """Возвращает (значения, вероятности).
    Строка 'p' -> значение = start + номер строки.
    Строка 'v p' -> значение берётся из файла."""
    xs, ps = [], []
    with open(filename) as f:
        for i, line in enumerate(f):
            parts = line.split()
            if not parts:
                continue
            if len(parts) == 1:
                xs.append(start + i)
                ps.append(float(parts[0]))
            else:
                xs.append(int(parts[0]))
                ps.append(float(parts[1]))
    return xs, ps

def plot(filename, title=None, start=1):
    xs, ps = load(filename, start)
    plt.figure(figsize=(10, 4))
    plt.bar(xs, ps, width=0.9)
    plt.xlabel("Значение")
    plt.ylabel("Вероятность")
    plt.title(title or filename)
    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.savefig("pictures/" + filename.split("/")[-1].rsplit(".", 1)[0] + ".png", dpi=150)
    plt.show()

plot("2_bonus.txt")
plot("2_pen.txt")
plot("2_simple.txt")
plot("2_triple.txt")
plot("3_double.txt")
plot("3_doublep.txt")
```
